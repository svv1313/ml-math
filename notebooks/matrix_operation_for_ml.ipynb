# **Matrix Operation for ML**

## Singular Value Decomposition (SVD)

```python
import torch
import numpy as np
```

```python
A = np.array([[-1, 2], [3, -2], [5, 7]])
A
```

```python
U, d, VT = np.linalg.svd(A)
```

```python
U
```

```python
d
```

```python
VT
```

```python
np.diag(d)
```

```python
D = np.concatenate((np.diag(d), [[0, 0]]), axis=0)
D
```

```python
np.dot(U, np.dot(D, VT))
```

```python
P = torch.tensor([[25, 2, -51], [3, -2, 1], [5, 7, 4.]])
```

```python
C, S, V = torch.svd(P)
```

```python
C
```

```python
S
```

```python
V
```

```python
PPt = torch.matmul(P, P.T)
PPt
```

```python
T_left_eigen = torch.linalg.eig(PPt)
Left_T = T_left_eigen.eigenvectors.real
```

```python
PtP = torch.matmul(P.T, P)
PtP
```

```python
T_right_eigen = torch.linalg.eig(PtP)
Right_T = T_right_eigen.eigenvectors.real
Right_T
```

```python
torch.sqrt(T_left_eigen.eigenvalues.real)
```

```python
torch.sqrt(T_right_eigen.eigenvalues.real)
```

```python
torch.isclose(torch.sqrt(T_left_eigen.eigenvalues.real), torch.sqrt(T_right_eigen.eigenvalues.real))
```
