Area Under the ROC curve
``` ROC - Receiver Operating Characteristic ```

When we don't have a function but we do have (x, y) coordinates, we can use the scikit-learn library's auc() method, which uses a numerical
approach (the trapezoidal rule) to find the area under the curve described by the coordinates:

````python
from sklearn.metrics import auc
````

````python
xs = [0, 0, 0.5, 0.5, 1]
ys = [0, 0.5, 0.5, 1, 1]
````

````python
auc(xs, ys)
````
