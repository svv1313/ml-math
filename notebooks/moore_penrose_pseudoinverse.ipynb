## The Moore-Penrose Pseudoinverse

```python
import torch
```

$A^+ = VD^+U^T$

```python
A = torch.tensor([[-1, 2], [3, -2], [5, 7.]])
```

```python
U, d, VT = torch.linalg.svd(A)
```

```python
V = VT.T
```

```python
d
```

```python
D = torch.diag(d)
D
```

```python
Dinv = torch.linalg.inv(D)
Dinv
```

```python
A.T
```

Dᴬ+ must have the same dimensions as A^T in order for V D+UT matrix multiplication to be possible:

```python
Dplus = torch.concatenate((Dinv, torch.tensor([[0, 0]]).T), axis=1)
Dplus
```

```python
Aplus = V @ Dplus @ U.T
```

```python
Aplus
```

Working out this derivation is helpful for understanding how Moore-Penrose pseudoinverses work, but unsurprisingly NumPy and PyTorch is
loaded with an existing method pinv ():

```python
torch.linalg.pinv(A)
```
