```python
import torch
import numpy as np
import matplotlib as plt
import math
```

```python
def f(my_x, my_y):
  return my_x**2 + my_y**2
```

```python
x = torch.tensor(0.).requires_grad_()
```

```python
y = torch.tensor(0.).requires_grad_()
```

```python
z = f(x, y)
```

```python
z.backward()
```

```python
x.grad
```

```python
y.grad
```

```python
def ex_for_class(my_x, my_y):
  return my_x**2 - my_y**2
```

```python
first_x = torch.tensor(3.).requires_grad_()
```

```python
first_y = torch.tensor(0.).requires_grad_()
```

```python
first_z = ex_for_class(first_x, first_y)
first_z
```

```python
first_z.backward()
print(first_x.grad)
print(first_y.grad)
```

```python
second_x = torch.tensor(-2.).requires_grad_()
second_y = torch.tensor(-3.).requires_grad_()
second_z = ex_for_class(second_x, second_y)
second_z
```

```python
second_z.backward()
print(second_x.grad)
print(second_y.grad)
```

```python
def cylinder_vol(my_r, my_l):
  return math.pi * my_r**2 * my_l
```

```python
r = torch.tensor(3.).requires_grad_()
```

```python
l = torch.tensor(5.).requires_grad_()
```

```python
v = cylinder_vol(r, l)
v
```

```python
v.backward()
```

```python
print(r.grad, l.grad)
```

```python
l.grad
```
