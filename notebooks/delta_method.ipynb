```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10, 10, 1000)
```

```python
def my_func(my_x):
  return my_x**2 + 2*my_x + 2
```

```python
y = my_func(x)
```

```python
fig, ax = plt.subplots()
plt.axvline(x=0, color='lightgray')
plt.axhline(y=0, color='lightgray')
ax.plot(x, y)
```

Let's identify the slope where, say, x = 2.

```python
p_x = 2
p_y = my_func(p_x)
```

```python
fig, ax = plt.subplots()
plt.axvline(x=0, color='lightgray')
plt.axhline(y=0, color='lightgray')
plt.scatter(p_x, p_y, color='red')
ax.plot(x, y)
```

```python
q_x = 5
```

```python
q_y= my_func(q_x)
```

```python
fig, ax = plt.subplots()
plt.axvline(x=0, color='lightgray')
plt.axhline(y=0, color='lightgray')
plt.scatter(p_x, p_y, color='red')
plt.scatter(q_x, q_y, color='yellow', zorder=3)
ax.plot(x, y)
```

```python
m = (q_y - p_y) / (q_x - p_x)
m
```

```python
b = p_y - m * p_x
b
```

```python
line_y = m * x + b
```

```python
fig, ax = plt.subplots()
plt.axvline(x=0, color='lightgray')
plt.axhline(y=0, color='lightgray')
plt.scatter(p_x, p_y, color='red')
plt.scatter(q_x, q_y, color='yellow', zorder=3)

plt.ylim(-5, 150)
plt.plot(x, line_y, c='orange')
ax.plot(x, y)
```

```python
x1 = -1
y1 = my_func(x1)
y1
```

```python
delta_x = 0.000001
```

```python
x2 = x1 + delta_x
y2 = my_func(x2)
y2
```

```python
fig, ax = plt.subplots()
plt.axvline(x=0, color='lightgray')
plt.axhline(y=0, color='lightgray')
plt.scatter(x1, y1, color='orange', zorder=3)
plt.scatter(x2, y2, color='red', zorder=3)
ax.plot(x, y)
```

```python
m = (y2 - y1) / (x2 - x1)
m
```

```python
b = y2 - m * x2
b
```

```python
line_y = m * x + b
```

```python
fig, ax = plt.subplots()
plt.axvline(x=0, color='lightgray')
plt.axhline(y=0, color='lightgray')
plt.scatter(x2, y2, color='red')
plt.scatter(x1, y1, color='orange', zorder=3)

plt.ylim(-5, 150)
plt.plot(x, line_y, c='orange')
ax.plot(x, y)
```
