```python
import torch
import matplotlib.pyplot as plt
```

```python
xs = torch.tensor([0., 1., 2., 3., 4., 5., 6., 7.])
ys = torch.tensor([11.86, 1.31, .62, .33, .09, -.67, -1.23, -1.37])
```

```python
def regression(my_x, my_m, my_b):
  return my_m * my_x + my_b
```

```python
m = torch.tensor([0.9]).requires_grad_()
b = torch.tensor([0.1]).requires_grad_()
```

```python
yhats = regression(xs, m, b)
yhats
```

```python
def mse(my_yhat, my_ys):
  sigma = torch.sum((my_yhat - my_ys) ** 2)
  return sigma / len(my_ys)
```

```python
C = mse(yhats, ys)
C
```

```python
C.backward()
```

```python
print(m.grad, b.grad)
```

```python
2 * 1 /len(ys) * torch.sum((yhats - ys) * xs)
```

```python
optimizer = torch.optim.SGD([m, b], lr=0.01)
```

```python
optimizer.step()
```

```python
C = mse(regression(xs, m, b), ys)
```

```python
print(m, b)
```

```python
epochs = 1000
for epoch in range(epochs):
  optimizer.zero_grad()
  yhats = regression(xs, m, b)
  C = mse(yhats, ys)
  C.backward()
  optimizer.step()
  print(m, b, C)
```
